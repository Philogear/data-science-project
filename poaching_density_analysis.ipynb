## Exercise 8.01 - Plotting Poaching Density Using Dot Density and Histograms

In this exercise we will use geoplotlib **dot**, **hist** and **voronoi** methods to visualize poaching incidents in Tanzania. The dataset contains 268 rows and 6 columns (*id_report*, *date_report*, *description*, *created_date*, *lat*, *lon*).

```python
# importing the necessary dependencies
import geoplotlib
from geoplotlib.utils import read_csv
```

```python
# loading the poaching dataset with geoplotlib's read_csv method
dataset = read_csv('../../Datasets/poaching_points_cleaned.csv')
```

```python
# looking at the dataset structure
dataset
```

The dataset is stored in a `DataAccessObject` class that is provided by geoplotlib. It does not have the same capabilities as a pandas `DataFrame` — it is meant for the simple and quick loading of data so that a visualization can be created. It gives a basic overview of what columns are present and how many rows the dataset has.

```python
# csv import with pandas
import pandas as pd
pd_dataset = pd.read_csv('../../Datasets/poaching_points_cleaned.csv')
pd_dataset.head()
```

### Dot density plot

```python
# plotting our dataset with points
geoplotlib.dot(dataset)
geoplotlib.show()
```

Looking at the rendered map, we can instantly see that some areas have more incidents than others. This insight could not have been easily identified by simply looking at the numbers in the dataset itself.

### Histogram layer

```python
# plotting our dataset as a histogram
geoplotlib.hist(dataset, binsize=20)
geoplotlib.show()
```

Histogram plots give us a better understanding of the density distribution of our dataset. There are some clear hotspots for poaching, and the areas without any poaching incidents are highlighted as well.

### Voronoi plot

```python
# plotting a voronoi map
geoplotlib.voronoi(dataset, cmap='Blues_r', \
                   max_area=1e5, alpha=255)
geoplotlib.show()
```

If we compare this Voronoi visualization with the histogram plot, one area draws a lot of attention: the center-right edge of the plot shows quite a large dark blue area with an even darker center — something that could easily have been overlooked with the histogram plot.
